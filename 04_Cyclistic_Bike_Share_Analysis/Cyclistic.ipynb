```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px

import glob
import os
```

```python
folder_path = "/content/drive/MyDrive/Cyclistic/data/raw"
```

```python
files = glob.glob(os.path.join(folder_path,"*.csv"))
print(files)
```

```python
df_list = []

for file in files:
  df = pd.read_csv(file)
  df_list.append(df)
```

```python
bike_data = pd.concat(df_list, ignore_index = True)
```

```python
bike_data.to_csv(
    "/content/drive/MyDrive/Cyclistic/data/cyclistic_combined.csv",index=False)
```

```python
missing = bike_data.isnull().sum().sort_values(ascending= False)
missing_percent = (missing / len(bike_data) * 100).round(2)

pd.DataFrame({
    "Missing Values": missing,
    "Percentage(%)": missing_percent
})
```

```python
bike_data[["started_at", "ended_at"]].dtypes
```

```python
bike_data["started_at"]=pd.to_datetime(bike_data["started_at"])
bike_data["ended_at"] =pd.to_datetime(bike_data["ended_at"])
```

```python
bike_data[["started_at","ended_at"]].dtypes
```

```python
bike_data["ride_length"] = (
    bike_data["ended_at"] - bike_data["started_at"]

).dt.total_seconds() /60
```

```python
bike_data.ride_length.head()
```

```python
bike_data = bike_data[bike_data["ride_length"]>0]
```

```python
bike_data["ride_length"].min()
```

Create Day of Week


```python
bike_data["day_of_week"] = bike_data["started_at"].dt.day_name()
```

```python
bike_data["day_of_week"].head()
```

```python
bike_data["month"] = bike_data["started_at"].dt.month_name()
bike_data["hour"] = bike_data["started_at"].dt.hour
bike_data["date"] = bike_data["started_at"].dt.date
```
